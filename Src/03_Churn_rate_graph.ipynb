Create a churn rate graph

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# 1. Load the dataset (using ../ since notebook is inside Src)
df = pd.read_csv("Data-raw/Telco-Customer-Churn-Cleaned.csv")

# 2. Calculate churn rate by contract type
df["Churn_numeric"] = df["Churn"].apply(lambda x: 1 if x == "Yes" else 0)
contract_churn = df.groupby("Contract")["Churn_numeric"].mean() * 100

# 3. Plotting
plt.figure(figsize=(8, 5))

sns.barplot(
    x=contract_churn.index, y=contract_churn.values, color="steelblue"
)

plt.title("Customer Churn Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Rate (%)")
plt.ylim(0, 100)

plt.show()
```
