```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(
    "../Data-raw/Telco-Customer-Churn-Cleaned.csv"
)

print(df.shape)
display(df.head())
```

Analyze MonthlyCharges


```python
print(df["MonthlyCharges"].describe())

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="MonthlyCharges",
    bins=30,
    kde=True
)

plt.title("Monthly Charges Distribution")
plt.xlabel("Monthly Charges")
plt.ylabel("Number of Customers")
plt.show()
```

Compare MonthlyCharges by Churn

```python
print(
    df.groupby("Churn")["MonthlyCharges"]
    .mean().round(2)
)

sns.boxplot(
    data=df,
    x="Churn",
    y="MonthlyCharges"
)

plt.title("Monthly Charges vs Churn")
plt.show()
```

Analyze TotalCharges

```python
print(df["TotalCharges"].describe())

sns.boxplot(
    data=df,
    x="Churn",
    y="TotalCharges"
)

plt.title("Total Charges vs Churn")
plt.show()
```

Compare average TotalCharges

```python
print(
    df.groupby("Churn")["TotalCharges"]
    .mean().round(2)
)
```
