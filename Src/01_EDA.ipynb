Count churned and retained customers

```python
print(df["Churn"].value_counts())
```

1.No: Customer is still using the service.
2.Yes: Customer has left the company.
3.value_counts(): Counts how many times each value     appears.

Calculate churn percentage

```python
churn_percentage = (
    df["Churn"].value_counts(normalize=True) * 100
)

print(churn_percentage.round(2))
```

Explanation: normalize=True calculates proportions instead of counts. Multiplying by 100 converts them into percentages.

Customer churn rate by contract type

Calculate churn rate

```python
contract_churn = (
    df.groupby("Contract")["Churn"]
    .apply(lambda x: (x == "Yes").mean() * 100)
)

print(contract_churn.round(2))
```
